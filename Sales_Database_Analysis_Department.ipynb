### Setup: Create Dummy Data and Initialize SQLite Database

I will create dummy CSV files for `Customers`, `Products`, and `Sales` tables, and then load them into an in-memory SQLite database. This will simulate the project's data environment.

```python
import pandas as pd
import sqlite3

# Create dummy data for Customers table
customers_data = {
    'CustomerID': [1, 2, 3, 4, 5],
    'CustomerName': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Gender': ['Female', 'Male', 'Male', 'Male', 'Female'],
    'City': ['New York', 'Los Angeles', 'Chicago', 'New York', 'Houston'],
    'Country': ['USA', 'USA', 'USA', 'USA', 'USA']
}
customers_df = pd.DataFrame(customers_data)
customers_df.to_csv('customers.csv', index=False)

# Create dummy data for Products table
products_data = {
    'ProductID': [101, 102, 103, 104, 105],
    'ProductName': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics'],
    'UnitPrice': [1200.00, 25.00, 75.00, 300.00, 50.00]
}
products_df = pd.DataFrame(products_data)
products_df.to_csv('products.csv', index=False)

# Create dummy data for Sales table
sales_data = {
    'SaleID': [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    'CustomerID': [1, 2, 1, 3, 4, 2, 5],
    'ProductID': [101, 102, 103, 101, 104, 105, 101],
    'Quantity': [1, 2, 1, 1, 1, 3, 1],
    'TotalAmount': [1200.00, 50.00, 75.00, 1200.00, 300.00, 150.00, 1200.00],
    'SaleDate': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-03', '2023-01-04', '2023-01-05', '2023-01-06']
}
sales_df = pd.DataFrame(sales_data)
sales_df.to_csv('sales.csv', index=False)

print("Dummy CSV files created: customers.csv, products.csv, sales.csv")

# Initialize an in-memory SQLite database
conn = sqlite3.connect(':memory:')

# Load data into SQLite tables
customers_df.to_sql('Customers', conn, index=False, if_exists='replace')
products_df.to_sql('Products', conn, index=False, if_exists='replace')
sales_df.to_sql('Sales', conn, index=False, if_exists='replace')

print("Data loaded into SQLite tables: Customers, Products, Sales")
```

### Helper Function to Run SQL Queries

This function will allow us to easily execute SQL queries and display the results in a pandas DataFrame.

```python
def run_sql_query(query):
    """Executes an SQL query and returns the result as a pandas DataFrame."""
    try:
        df = pd.read_sql_query(query, conn)
        return df
    except Exception as e:
        print(f"Error executing query: {e}")
        return None
```

SQLite uses `PRAGMA table_info(tablename)` to get schema information. Primary keys are indicated by `pk = 1`. Foreign keys are not explicitly marked as such in `table_info` output, but are identified by column names like `CustomerID` and `ProductID` in the `Sales` table, linking to the `Customers` and `Products` tables respectively.

```python
print("\n--- Customers Table Schema ---")
display(run_sql_query("PRAGMA table_info(Customers);"))

print("\n--- Products Table Schema ---")
display(run_sql_query("PRAGMA table_info(Products);"))

print("\n--- Sales Table Schema ---")
display(run_sql_query("PRAGMA table_info(Sales);"))
```

### Activity 5 – CASE Statements

**Task:**
Create calculated columns using `CASE`. Examples:
*   Sales Category
    *   High
    *   Medium
    *   Low
*   Customer Type
    *   Premium
    *   Regular

#### Create a 'Sales Category' column based on `TotalAmount`

```python
query_sales_category = """
SELECT
    SaleID,
    TotalAmount,
    CASE
        WHEN TotalAmount >= 1000 THEN 'High'
        WHEN TotalAmount >= 100 THEN 'Medium'
        ELSE 'Low'
    END AS SalesCategory
FROM Sales;
"""
display(run_sql_query(query_sales_category))
```

#### Create a 'Customer Type' column based on `TotalAmount` (simplified example, usually based on total historical purchases or other criteria)

```python
query_customer_type = """
SELECT
    c.CustomerID,
    c.CustomerName,
    SUM(s.TotalAmount) AS TotalPurchases,
    CASE
        WHEN SUM(s.TotalAmount) >= 1500 THEN 'Premium'
        ELSE 'Regular'
    END AS CustomerType
FROM Customers c
JOIN Sales s ON c.CustomerID = s.CustomerID
GROUP BY c.CustomerID, c.CustomerName;
"""
display(run_sql_query(query_customer_type))
```

### Activity 9 – GROUP BY Analysis

**Tasks:**
Write queries to analyze:
*   Total sales by city
*   Total sales by product category
*   Number of customers by country
*   Average sales by customer
*   Product sales summary

#### Total sales by city

```python
query_total_sales_by_city = """
SELECT
    c.City,
    SUM(s.TotalAmount) AS TotalSales
FROM Sales s
JOIN Customers c ON s.CustomerID = c.CustomerID
GROUP BY c.City;
"""
display(run_sql_query(query_total_sales_by_city))
```

#### Total sales by product category

```python
query_total_sales_by_category = """
SELECT
    p.Category,
    SUM(s.TotalAmount) AS TotalSales
FROM Sales s
JOIN Products p ON s.ProductID = p.ProductID
GROUP BY p.Category;
"""
display(run_sql_query(query_total_sales_by_category))
```

#### Number of customers by country

```python
query_customers_by_country = "SELECT Country, COUNT(CustomerID) AS NumberOfCustomers FROM Customers GROUP BY Country;"
display(run_sql_query(query_customers_by_country))
```

#### Average sales by customer

```python
query_avg_sales_by_customer = """
SELECT
    c.CustomerName,
    AVG(s.TotalAmount) AS AverageSale
FROM Sales s
JOIN Customers c ON s.CustomerID = c.CustomerID
GROUP BY c.CustomerName;
"""
display(run_sql_query(query_avg_sales_by_customer))
```

#### Product sales summary (Total Quantity Sold, Total Revenue by Product)

```python
query_product_sales_summary = """
SELECT
    p.ProductName,
    SUM(s.Quantity) AS TotalQuantitySold,
    SUM(s.TotalAmount) AS TotalRevenue
FROM Sales s
JOIN Products p ON s.ProductID = p.ProductID
GROUP BY p.ProductName
ORDER BY TotalRevenue DESC;
"""
display(run_sql_query(query_product_sales_summary))
```

### Activity 11 – Sales Performance Table

**Task:**
Prepare the following report using SQL:
*   Product Category
*   Total Sales
*   Total Orders
*   Average Sale
*   Highest Sale

Save the output.

#### Generate Sales Performance Report by Product Category

```python
query_sales_performance_report = """
SELECT
    p.Category AS "Product Category",
    SUM(s.TotalAmount) AS "Total Sales",
    COUNT(s.SaleID) AS "Total Orders",
    AVG(s.TotalAmount) AS "Average Sale",
    MAX(s.TotalAmount) AS "Highest Sale"
FROM Sales s
JOIN Products p ON s.ProductID = p.ProductID
GROUP BY p.Category
ORDER BY "Total Sales" DESC;
"""
display(run_sql_query(query_sales_performance_report))
```

### Activity 12 – Business Questions

**Tasks:**
Answer the following using SQL:
1.  Which customer generated the highest sales?
2.  Which city has the highest revenue?
3.  Which product category performs best?
4.  Which products are rarely purchased?
5.  Which month recorded the highest sales?
6.  What is the average order value?
7.  Which customers made multiple purchases?
8.  Which products generated the highest revenue?

Document both the SQL query and the business insight.

#### 1. Which customer generated the highest sales?

```python
query_highest_sales_customer = """
SELECT
    c.CustomerName,
    SUM(s.TotalAmount) AS TotalSales
FROM Sales s
JOIN Customers c ON s.CustomerID = c.CustomerID
GROUP BY c.CustomerName
ORDER BY TotalSales DESC
LIMIT 1;
"""
display(run_sql_query(query_highest_sales_customer))
```

#### 2. Which city has the highest revenue?

```python
query_highest_revenue_city = """
SELECT
    c.City,
    SUM(s.TotalAmount) AS TotalRevenue
FROM Sales s
JOIN Customers c ON s.CustomerID = c.CustomerID
GROUP BY c.City
ORDER BY TotalRevenue DESC
LIMIT 1;
"""
display(run_sql_query(query_highest_revenue_city))
```

#### 3. Which product category performs best (highest total sales)?

```python
query_best_product_category = """
SELECT
    p.Category,
    SUM(s.TotalAmount) AS TotalSales
FROM Sales s
JOIN Products p ON s.ProductID = p.ProductID
GROUP BY p.Category
ORDER BY TotalSales DESC
LIMIT 1;
"""
display(run_sql_query(query_best_product_category))
```

#### 4. Which products are rarely purchased (e.g., purchased only once or lowest quantity)?

```python
query_rarely_purchased_products = """
SELECT
    p.ProductName,
    COUNT(s.SaleID) AS PurchaseCount,
    SUM(s.Quantity) AS TotalQuantitySold
FROM Sales s
JOIN Products p ON s.ProductID = p.ProductID
GROUP BY p.ProductName
ORDER BY TotalQuantitySold ASC, PurchaseCount ASC
LIMIT 3; -- Display top 3 least purchased products
"""
display(run_sql_query(query_rarely_purchased_products))
```

#### 5. Which month recorded the highest sales?

```python
query_highest_sales_month = """
SELECT
    STRFTIME('%Y-%m', SaleDate) AS SalesMonth,
    SUM(TotalAmount) AS MonthlySales
FROM Sales
GROUP BY SalesMonth
ORDER BY MonthlySales DESC
LIMIT 1;
"""
display(run_sql_query(query_highest_sales_month))
```

#### 6. What is the average order value?

```python
query_average_order_value = "SELECT AVG(TotalAmount) AS AverageOrderValue FROM Sales;"
display(run_sql_query(query_average_order_value))
```

#### 7. Which customers made multiple purchases?

```python
query_multiple_purchases_customers = """
SELECT
    c.CustomerName,
    COUNT(s.SaleID) AS NumberOfPurchases
FROM Sales s
JOIN Customers c ON s.CustomerID = c.CustomerID
GROUP BY c.CustomerName
HAVING COUNT(s.SaleID) > 1
ORDER BY NumberOfPurchases DESC;
"""
display(run_sql_query(query_multiple_purchases_customers))
```

#### 8. Which products generated the highest revenue?

```python
query_highest_revenue_products = """
SELECT
    p.ProductName,
    SUM(s.TotalAmount) AS TotalRevenue
FROM Sales s
JOIN Products p ON s.ProductID = p.ProductID
GROUP BY p.ProductName
ORDER BY TotalRevenue DESC
LIMIT 1;
"""
display(run_sql_query(query_highest_revenue_products))
```

### Activity 13 – SQL Challenge

**Task:**
Write a single SQL query to produce a management report containing:
*   Customer Name
*   City
*   Product Category
*   Number of Orders
*   Total Sales
*   Average Purchase
*   Customer Status (Premium/Regular using `CASE`)

Sort the report by Total Sales in descending order.

#### Generate Management Report

```python
query_management_report = """
SELECT
    c.CustomerName,
    c.City,
    p.Category AS ProductCategory,
    COUNT(DISTINCT s.SaleID) AS NumberOfOrders,
    SUM(s.TotalAmount) AS TotalSales,
    AVG(s.TotalAmount) AS AveragePurchase,
    CASE
        WHEN SUM(s.TotalAmount) >= 1500 THEN 'Premium'
        ELSE 'Regular'
    END AS CustomerStatus
FROM Sales s
JOIN Customers c ON s.CustomerID = c.CustomerID
JOIN Products p ON s.ProductID = p.ProductID
GROUP BY c.CustomerID, c.CustomerName, c.City, p.Category
ORDER BY TotalSales DESC;
"""
display(run_sql_query(query_management_report))
```
